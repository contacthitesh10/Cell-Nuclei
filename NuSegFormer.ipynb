```python
# ============================================================================
# ViT-UNet 
# ============================================================================

import os, zipfile, cv2, warnings
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, mixed_precision
from glob import glob
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.model_selection import KFold
import albumentations as A

warnings.filterwarnings("ignore")
tf.get_logger().setLevel("ERROR")

mixed_precision.set_global_policy('float32')

# ============================================================================
# SECTION 1: CONFIGURATION
# ============================================================================

IMG_SIZE   = 512
EMBED_DIM  = 256
NUM_HEADS  = 8
MLP_DIM    = 512
NUM_LAYERS = 4
DROPOUT    = 0.1
BATCH_SIZE = 8
EPOCHS     = 40
K_FOLDS    = 5
SEED       = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

try:
    gpus = tf.config.list_physical_devices('GPU')
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)
except RuntimeError:
    pass

# ============================================================================
# SECTION 2: DATASET EXTRACTION
# ============================================================================

zip_path     = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

print("=" * 70)
print("STEP 1: Extracting dataset...")
print("=" * 70)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(extract_root)
    print("  Extraction complete.")
else:
    print("  Already extracted, skipping.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
data_path  = candidates[0] if len(candidates) == 1 else extract_root
print(f"  Dataset folder: {data_path}")

# ============================================================================
# SECTION 3: DATA LOADING
# ============================================================================

def load_data(data_dir, img_size=(IMG_SIZE, IMG_SIZE)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"  Found {len(all_ids)} samples")

    for img_id in tqdm(all_ids, desc="  Loading"):
        img_paths  = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks",  "*.png"))
        if not img_paths or not mask_paths:
            continue

        image = cv2.imread(img_paths[0])
        if image is None:
            continue
        image = cv2.resize(image, img_size).astype(np.float32) / 255.0

        merged = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            merged = np.maximum(merged, cv2.resize(m, img_size))

        if merged.sum() == 0:
            continue

        mask = (merged > 127).astype(np.float32)[..., np.newaxis]
        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)


print("\n" + "=" * 70)
print("STEP 2: Loading data...")
print("=" * 70)

X, y = load_data(data_path)
print(f"  Images: {X.shape}  |  Masks: {y.shape}")

# ============================================================================
# SECTION 4: AUGMENTATION
# Original + 5 augmentations applied individually — final set is N x 6.
# ============================================================================

augmentations = [
    A.HorizontalFlip(p=1.0),
    A.VerticalFlip(p=1.0),
    A.GaussianBlur(blur_limit=(3, 3), sigma_limit=2, p=1.0),
    A.GaussianBlur(blur_limit=(7, 7), sigma_limit=4, p=1.0),
    A.Rotate(limit=270, interpolation=cv2.INTER_LINEAR,
             border_mode=cv2.BORDER_REFLECT_101, p=1.0),
]


def augment_dataset(X_train, y_train):
    aug_X = [X_train.copy()]
    aug_y = [y_train.copy()]

    for aug_idx, aug in enumerate(augmentations):
        print(f"  Aug {aug_idx + 1}/{len(augmentations)}: {aug.__class__.__name__}...")
        batch_X, batch_y = [], []
        for img, mask in zip(X_train, y_train):
            result = aug(
                image=(img  * 255).astype(np.uint8),
                mask =(mask * 255).astype(np.uint8).squeeze()
            )
            batch_X.append(result["image"].astype(np.float32) / 255.0)
            batch_y.append(
                (result["mask"] > 127).astype(np.float32)[..., np.newaxis]
            )
        aug_X.append(np.stack(batch_X))
        aug_y.append(np.stack(batch_y))

    X_out = np.concatenate(aug_X, axis=0).astype(np.float32)
    y_out = np.concatenate(aug_y, axis=0).astype(np.float32)
    print(f"  Final training set: {X_out.shape}  ({len(X_train)} x 6)")
    return X_out, y_out

# ============================================================================
# SECTION 5: MODEL ARCHITECTURE — Hybrid UNet + ViT Bottleneck (5-Gate Attention)
# ============================================================================

#-- Used in Decoder --
def conv_block(x, filters):
    x = layers.Conv2D(filters, 3, padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)
    x = layers.Dropout(DROPOUT)(x)
    x = layers.Conv2D(filters, 3, padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)
    return x

# -- Decoder Upsampling Step --
def upconv_block(x, skip, filters):
    x = layers.Conv2DTranspose(filters, 2, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip])
    x = conv_block(x, filters)
    return x

# -- Adds a learned position embedding to the sequence of patch tokens --
class AddPositionalEmbedding(layers.Layer):
    """Learned positional embedding via add_weight — avoids Keras 3
    Embedding layer cast errors on the TF backend."""
    def __init__(self, n_patches, dim, **kwargs):
        super().__init__(**kwargs)
        self.n_patches = n_patches
        self.dim       = dim

    def build(self, input_shape):
        self.pos_emb = self.add_weight(
            name="pos_emb",
            shape=(1, self.n_patches, self.dim),
            initializer="random_normal",
            trainable=True,
        )
        super().build(input_shape)

    def call(self, x):
        return x + tf.cast(self.pos_emb, dtype=x.dtype)


# Gate 1: Query input gate
# Gate 2: Key input gate
# Gate 3: Value input gate
# Gate 4: SDPA output gate
# Gate 5: Dense output gate

class GatedMultiHeadAttention(layers.Layer):
    def __init__(self, num_heads, key_dim, name_suffix="", **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = num_heads * key_dim

        self.attention = layers.MultiHeadAttention(
            num_heads=num_heads, key_dim=key_dim,
            name=f"mha_{name_suffix}"
        )

        def _gate(tag):
            return layers.Dense(
                self.embed_dim, activation="sigmoid",
                name=f"{tag}_gate_{name_suffix}"
            )

        self.query_gate = _gate("query")
        self.key_gate   = _gate("key")
        self.value_gate = _gate("value")
        self.sdpa_gate  = _gate("sdpa")
        self.dense_gate = _gate("dense")

        self.query_mul  = layers.Multiply(name=f"qmul_{name_suffix}")
        self.key_mul    = layers.Multiply(name=f"kmul_{name_suffix}")
        self.value_mul  = layers.Multiply(name=f"vmul_{name_suffix}")
        self.sdpa_mul   = layers.Multiply(name=f"smul_{name_suffix}")
        self.dense_mul  = layers.Multiply(name=f"dmul_{name_suffix}")

    def call(self, x, training=None):
        orig = x
        q   = self.query_mul([x, self.query_gate(x)])
        k   = self.key_mul  ([x, self.key_gate(x)])
        v   = self.value_mul([x, self.value_gate(x)])
        out = self.attention(query=q, key=k, value=v, training=training)
        out = self.sdpa_mul ([out, self.sdpa_gate(orig)])
        out = self.dense_mul([out, self.dense_gate(orig)])
        return out


# ============================================================================
# ReGLU feed-forward block
# This replaces the standard GELU MLP
# ============================================================================

class ReGLUFeedForward(layers.Layer):
    """
    Projects to 2 * MLP_DIM, splits into (value, gate), then:
        output = value * ReLU(gate)
    followed by a down-projection back to EMBED_DIM.
    """
    def __init__(self, embed_dim, mlp_dim, dropout=0.0, **kwargs):
        super().__init__(**kwargs)
        # Project up to 2x width so the gated output has width == mlp_dim
        self.up_proj   = layers.Dense(mlp_dim * 2, use_bias=True)
        self.down_proj = layers.Dense(embed_dim,   use_bias=True)
        self.drop1     = layers.Dropout(dropout)
        self.drop2     = layers.Dropout(dropout)

    def call(self, x, training=None):
        projected = self.up_proj(x)                          # (..., 2 * mlp_dim)
        value, gate = tf.split(projected, 2, axis=-1)        # each (..., mlp_dim)
        x = value * tf.nn.relu(gate)                         # ReGLU activation
        x = self.drop1(x, training=training)
        x = self.down_proj(x)                                # (..., embed_dim)
        x = self.drop2(x, training=training)
        return x


def transformer_block(x, block_idx):
    residual = x
    attn = GatedMultiHeadAttention(
        num_heads=NUM_HEADS,
        key_dim=EMBED_DIM // NUM_HEADS,
        name_suffix=f"blk{block_idx}"
    )(x)
    x = layers.Add()([residual, attn])
    x = layers.LayerNormalization(epsilon=1e-6)(x)

    residual = x
    # ---- ReGLU replaces the previous Dense(GELU) -> Dense stack ----
    mlp = ReGLUFeedForward(
        embed_dim=EMBED_DIM,
        mlp_dim=MLP_DIM,
        dropout=DROPOUT,
        name=f"reglu_ffn_blk{block_idx}"
    )(x)
    x = layers.Add()([residual, mlp])
    x = layers.LayerNormalization(epsilon=1e-6)(x)
    return x


def vit_bottleneck(x, n_layers=NUM_LAYERS):
    _, H, W, C = x.shape

    if C != EMBED_DIM:
        x = layers.Conv2D(EMBED_DIM, 1, padding="same", name="bottleneck_proj")(x)
        C = EMBED_DIM

    n_patches = H * W
    flat = layers.Reshape((n_patches, EMBED_DIM))(x)
    flat = AddPositionalEmbedding(n_patches, EMBED_DIM, name="pos_emb")(flat)

    for i in range(n_layers):
        flat = transformer_block(flat, block_idx=i)

    out = layers.Reshape((H, W, EMBED_DIM))(flat)
    out = layers.Conv2D(C, 1, padding="same", name="bottleneck_unproj")(out)
    return out


def build_hybrid_unet_vit():
    inputs = layers.Input((IMG_SIZE, IMG_SIZE, 3), name="input_image")

    # ---- EfficientNetV2-S encoder (ImageNet pretrained) ----
    # Input is normalized to [0,1] — EfficientNetV2 expects this natively
    effnet = tf.keras.applications.EfficientNetV2S(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    # Skip connections at 4 scales
    # block1b_add  -> 256x256 x  48  (stride 1 after stem)
    # block2d_add  -> 128x128 x  64
    # block4a_expand_activation -> 64x64 x 160
    # block6a_expand_activation -> 32x32 x 272
    # top_activation             -> 16x16 x1280  (bottleneck input)
    e1 = effnet.get_layer("block1b_add").output                   # 256x256 x  48
    e2 = effnet.get_layer("block2d_add").output                   # 128x128 x  64
    e3 = effnet.get_layer("block4a_expand_activation").output     #  64x64  x 160
    e4 = effnet.get_layer("block6a_expand_activation").output     #  32x32  x 272
    p4 = effnet.get_layer("top_activation").output                #  16x16  x1280

    bottleneck = vit_bottleneck(p4)

    d4 = upconv_block(bottleneck, e4, 272)
    d3 = upconv_block(d4,         e3, 160)
    d2 = upconv_block(d3,         e2,  64)
    d1 = upconv_block(d2,         e1,  48)

    # Final upsample to restore full 512x512 resolution
    # (EfficientNetV2S stem halves input to 256x256)
    d0 = layers.Conv2DTranspose(32, 2, strides=2, padding="same")(d1)
    d0 = layers.Conv2D(32, 3, padding="same", use_bias=False)(d0)
    d0 = layers.BatchNormalization()(d0)
    d0 = layers.Activation("relu")(d0)

    # Output layer
    x = layers.Conv2D(1, 1, name="pre_output")(d0)
    outputs = layers.Activation("sigmoid", name="output")(x)

    return models.Model(inputs=inputs, outputs=outputs, name="Hybrid_UNet_ViT_EfficientNetV2S_ReGLU")

# ============================================================================
# SECTION 6: LOSS FUNCTIONS & METRICS
# ============================================================================

def dice_loss(y_true, y_pred):
    smooth = 1e-6
    y_true = tf.cast(tf.reshape(y_true, [-1]), tf.float32)
    y_pred = tf.cast(tf.reshape(y_pred, [-1]), tf.float32)
    inter  = tf.reduce_sum(y_true * y_pred)
    return 1.0 - (2.0 * inter + smooth) / (
        tf.reduce_sum(y_true) + tf.reduce_sum(y_pred) + smooth
    )


def combined_loss(y_true, y_pred):
    """Dice + BCE, equal weighting (1.0 each)."""
    return dice_loss(y_true, y_pred) + tf.keras.losses.binary_crossentropy(
        tf.cast(y_true, tf.float32), tf.cast(y_pred, tf.float32)
    )


def dice_score_np(y_true, y_pred, threshold=0.5):
    y_pred = (y_pred > threshold).astype(np.float32)
    y_true = y_true.astype(np.float32)
    inter  = np.sum(y_true * y_pred)
    return (2.0 * inter + 1e-6) / (np.sum(y_true) + np.sum(y_pred) + 1e-6)


def iou_score_np(y_true, y_pred, threshold=0.5):
    y_pred = (y_pred > threshold).astype(np.float32)
    y_true = y_true.astype(np.float32)
    inter  = np.sum(y_true * y_pred)
    union  = np.sum(y_true) + np.sum(y_pred) - inter
    return (inter + 1e-6) / (union + 1e-6)

# ============================================================================
# SECTION 7: K-FOLD CROSS-VALIDATION TRAINING
# ============================================================================

print("\n" + "=" * 70)
print("STEP 3: K-Fold Cross-Validation Training")
print("=" * 70)

kf           = KFold(n_splits=K_FOLDS, shuffle=True, random_state=SEED)
fold_results = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X), 1):
    print(f"\n{'─' * 70}")
    print(f"  FOLD {fold} / {K_FOLDS}")
    print(f"{'─' * 70}")

    X_train_raw, X_val = X[train_idx], X[val_idx]
    y_train_raw, y_val = y[train_idx], y[val_idx]

    tf.keras.backend.clear_session()

    X_train, y_train = augment_dataset(X_train_raw, y_train_raw)

    model = build_hybrid_unet_vit()

    if fold == 1:
        print("\n  Model Summary:")
        model.summary(line_length=100)

    steps_per_epoch = max(1, len(X_train) // BATCH_SIZE)
    total_steps     = steps_per_epoch * EPOCHS

    lr_schedule = tf.keras.optimizers.schedules.CosineDecayRestarts(
        initial_learning_rate=3e-4,
        first_decay_steps=total_steps // 2,
        t_mul=1.5,
        m_mul=0.9,
        alpha=1e-6
    )

    model.compile(
        optimizer=tf.keras.optimizers.AdamW(
            learning_rate=lr_schedule,
            weight_decay=1e-4
        ),
        loss=combined_loss,
        metrics=["accuracy"]
    )

    ckpt_path = f"/content/best_weights_fold{fold}.weights.h5"

    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            filepath=ckpt_path,
            monitor="val_loss",
            mode="min",
            save_best_only=True,
            save_weights_only=True,
            verbose=1
        ),
        tf.keras.callbacks.TerminateOnNaN(),
    ]

    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=callbacks,
        verbose=1
    )

    model.load_weights(ckpt_path)
    print(f"  Loaded best weights from: {ckpt_path}")

    preds    = model.predict(X_val, batch_size=BATCH_SIZE, verbose=0)
    dice     = dice_score_np(y_val, preds, threshold=0.5)
    iou      = iou_score_np(y_val,  preds, threshold=0.5)
    best_ep  = int(np.argmin(history.history["val_loss"]))
    val_loss = history.history["val_loss"][best_ep]
    val_acc  = history.history["val_accuracy"][best_ep]

    fold_results.append({
        "fold":     fold,
        "val_loss": val_loss,
        "val_acc":  val_acc,
        "dice":     dice,
        "iou":      iou,
        "history":  history.history,
        "X_val":    X_val,
        "y_val":    y_val,
        "preds":    preds,
        "ckpt":     ckpt_path,
    })

    print(f"\n  Fold {fold} Results:")
    print(f"     Weights saved : {ckpt_path}")
    print(f"     Val Loss      : {val_loss:.4f}")
    print(f"     Val Accuracy  : {val_acc:.4f}")
    print(f"     Dice Score    : {dice:.4f}")
    print(f"     IoU Score     : {iou:.4f}")

    # Free augmented training data — not needed after fold completes
    del X_train, y_train
    import gc; gc.collect()

# ============================================================================
# SECTION 8: FINAL RESULTS REPORT
# ============================================================================

print("\n" + "=" * 70)
print("FINAL CROSS-VALIDATION RESULTS")
print("=" * 70)

dice_arr = np.array([r["dice"]     for r in fold_results])
iou_arr  = np.array([r["iou"]      for r in fold_results])
acc_arr  = np.array([r["val_acc"]  for r in fold_results])
loss_arr = np.array([r["val_loss"] for r in fold_results])

print(f"\n{'Fold':<6} {'Val Loss':<13} {'Val Acc':<13} {'Dice@0.50':<13} {'IoU@0.50':<13} {'Weights'}")
print("-" * 95)
for r in fold_results:
    print(f"  {r['fold']:<4} {r['val_loss']:<13.4f} {r['val_acc']:<13.4f} "
          f"{r['dice']:<13.4f} {r['iou']:<13.4f} {r['ckpt']}")

print("-" * 95)
print(f"  {'MEAN':<4} {loss_arr.mean():<13.4f} {acc_arr.mean():<13.4f} {dice_arr.mean():<13.4f} {iou_arr.mean():<13.4f}")
print(f"  {'STD':<4} {loss_arr.std():<13.4f} {acc_arr.std():<13.4f} {dice_arr.std():<13.4f} {iou_arr.std():<13.4f}")
print("=" * 70)
print(f"\n  Mean Dice = {dice_arr.mean():.4f} +/- {dice_arr.std():.4f}")
print(f"  Mean IoU  = {iou_arr.mean():.4f} +/- {iou_arr.std():.4f}")

def plot_training_curves(fold_results):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle("Training Curves — All Folds (ReGLU)", fontsize=14, fontweight="bold")

    for r in fold_results:
        h    = r["history"]
        fold = r["fold"]
        ep   = range(1, len(h["loss"]) + 1)

        axes[0].plot(ep, h["loss"],         label=f"Fold {fold} train")
        axes[0].plot(ep, h["val_loss"],     linestyle="--", label=f"Fold {fold} val")
        axes[1].plot(ep, h["accuracy"],     label=f"Fold {fold} train")
        axes[1].plot(ep, h["val_accuracy"], linestyle="--", label=f"Fold {fold} val")

    axes[0].set_title("Loss");     axes[0].set_xlabel("Epoch"); axes[0].legend(fontsize=7)
    axes[1].set_title("Accuracy"); axes[1].set_xlabel("Epoch"); axes[1].legend(fontsize=7)

    plt.tight_layout()
    plt.savefig("/content/training_curves.png", dpi=150, bbox_inches="tight")
    print("  Saved: /content/training_curves.png")

plot_training_curves(fold_results)

print("\n  Pipeline complete.")
```
